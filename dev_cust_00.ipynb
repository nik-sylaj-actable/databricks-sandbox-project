```python
# Create catalog and schema if they do not exist
spark.sql("CREATE CATALOG IF NOT EXISTS dev")
spark.sql("CREATE SCHEMA IF NOT EXISTS dev.bronze")

# Read the sample customer dataset
df = spark.read.table("samples.tpch.customer")
```

```python
df.show(10)
```

```python
display(df.filter((df.c_mktsegment == "BUILDING") | (df.c_mktsegment == "MACHINERY")))
```

```python
df = df.filter((df.c_mktsegment == "BUILDING") | (df.c_mktsegment == "MACHINERY"))
```

```python
# Write out as a managed table in dev.bronze
df.write.mode("overwrite").saveAsTable("dev.bronze.customer")
```
